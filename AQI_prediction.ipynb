**Selected Problem Domain** - Environmental Health Monitoring (Air Quality Index Analysis and Prediction)

Air pollution is a growing global concern that poses serious risks to public health and the environment. This work applies machine learning techniques to historical Air Quality Index (AQI) data to analyse patterns and predict pollution levels based on measured environmental variables. The aim is to evaluate and compare different AI techniques in their ability to accurately estimate AQI values, supporting more effective environmental health monitoring

#### **Environmental Health Monitoring**

Air pollution has emerged as one of the most serious threats to human health and environmental sustainability worldwide. According to reports by the World Health Organization, millions of premature deaths each year are linked to prolonged exposure to polluted air, making air pollution the single greatest environmental risk to human well-being (Dao et al, 2022). Harmful pollutants released through industrial activities, fossil fuel combustion, vehicular emissions, and agricultural practices contribute significantly to respiratory diseases, cardiovascular disorders, and other long-term health complications (Tırınk,2025).

To assess and communicate the severity of pollution, continuous monitoring and accurate prediction of AQI are therefore essential for protecting public health, enabling early warning systems, and supporting policymakers in implementing effective mitigation strategies (Hardini et al, 2023). With the increasing availability of environmental data, Artificial Intelligence and Machine Learning techniques have become critical tools for improving the reliability and timeliness of air quality analysis and prediction (Binbusayyis et al., 2024).

#### **Air Quality Indexing(AQI)**

The Air Quality Index (AQI) is a standardized numerical scale used to represent air pollution levels and their potential short-term health impacts in a clear and interpretable manner (Gupta et al.,2023). By aggregating information from multiple pollutants into a single index value, AQI enables real-time assessment of ambient air quality and supports timely public health communication . Higher AQI values indicate increasingly polluted air and are associated with greater health risks, whereas lower values reflect cleaner air conditions.

AQI calculation is primarily based on measured concentrations of key air pollutants, including particulate matter (PM2.5 and PM10) and gaseous pollutants such as nitrogen dioxide (NO₂), sulfur dioxide (SO₂), carbon monoxide (CO), and ozone (O₃) (Gupta et al., 2023). In addition to pollutant concentrations, meteorological variables such as temperature, humidity, wind speed, and atmospheric pressure are often recorded alongside air quality measurements, as they influence pollutant dispersion and accumulation. Each pollutant concentration is first converted into a sub-index, and the overall AQI value is determined by the maximum of these sub-indices. #### **AI in AQI Prediction**


Air quality monitoring has greatly benefited from the adoption of artificial learning (ML) and deep learning (DL) techniques which overcome the limitations of traditional statistical methods in handling non-linear and complex environmental data (Binbusayyis et al., 2024). Early approaches relied on linear models like Linear Regression (LR) have been applied to quantify relationships between pollutants and AQI however, their predictive capacity is often limited, particularly in the presence of sudden environmental changes (Sharma, 2023). More robust techniques, such as Support Vector Regression (SVR) and Random Forest Regression (RFR), have demonstrated higher accuracy and resilience against overfitting, with RFR effectively modeling non-linear relationships in large datasets (Gupta et al., 2023).

Ensemble models, especially XGBoost and LightGBM, have emerged as high-performing approaches due to their ability to capture complex feature interactions and scale efficiently with data  (Tırınk,2025). Hybrid and stacked architectures, combining models like RFR and XGBoost, further enhance performance by mitigating the shortcomings of individual algorithms (Sharma et al., 2021).

Deep learning models, including Artificial Neural Networks (ANNs), Multilayer Perceptrons (MLPs), and Recurrent Neural Networks (RNNs), have also been employed for sequential and time-series pollutant data.  Additionally, models like Sparse Spectrum Gaussian Process Regression (SSGPR) allow real-time prediction while modeling uncertainty (Hardini et al., 2023). Overall, AI techniques provide a significant improvement in the accuracy, efficiency, and scalability of AQI prediction over traditional methods..


#### **Methodological Approaches and Evaluation Metrics**

Effective AQI prediction relies on comprehensive methodological strategies, including rigorous data pre-processing, model optimization, and standardized evaluation metrics. Pre-processing steps address challenges such as missing values, using methods like Deep GAN imputation (Binbusayyis et al., 2024), and imbalanced datasets, tackled with SMOTE (Gupta et al., 2023) to prevent biased predictions. Feature scaling, normalization, and exploratory data analysis ensure that relevant pollutant and environmental variables contribute meaningfully to model learning.

Model training typically employs k-fold cross-validation to evaluate generalization and hyperparameter tuning, often via GridSearchCV, to optimize model performance. Ensemble techniques, such as bagging and stacking, further enhance prediction stability and accuracy. Evaluation metrics for regression models include Root Mean Squared Error (RMSE), Mean Absolute Error (MAE), Mean Squared Error (MSE), and R², which quantify predictive accuracy, error magnitude, and model fit. For classification tasks, metrics like Accuracy, F1 Score, and Matthew’s Correlation Coefficient (MCC) are used to validate categorical predictions, ensuring robust assessment across diverse AQI datasets (Sharma et al., 2021).

#### **Future Improvements & Challenges**

AQI prediction faces significant challenges, particularly due to limited spatial coverage of monitoring stations and frequent data gaps, which can bias model outcomes and reduce dataset reliability (Anggraini et al.; Binbusayyis et al.). Many models also struggle to generalize beyond short-term or localized datasets and to capture diurnal or seasonal variations in pollutant levels (Tırınk; Gupta et al.). Future improvements emphasize integrating auxiliary meteorological and pollutant source data, applying advanced imputation techniques, and developing scalable AI and deep learning models capable of providing region-wide, long-term predictions, thereby enhancing accuracy, reliability, and actionable insights for policy and planning (Sharma et al.; Hardini et al.).

# Comparison Of Techniques
---

For the task of predicting the Air Quality Index (AQI) in this study, three machine learning techniques have been selected for comparison:
1. XGBoost
2. Random Forest Regression
3. Neural Network

Each technique is evaluated based on its strengths and weaknesses, suitability for AQI prediction, input requirements, and expected output. The comparison highlights why XGBoost and Random Forest are chosen ensembling in the prototype implementation in Part C.

**XGBoost**

* **Strengths**

    * Uses gradient boosting to iteratively reduce prediction errors, capturing complex non-linear relationships in pollutant data.
    * Handles sparse or missing data efficiently, which is common in AQI datasets.
    * L1 and L2 regularization reduce overfitting, improving generalization to unseen data.
    * Optimized for parallel computation, making it faster than many other boosting methods.

* **Weaknesses**

    * Computationally intensive compared to simpler models.
    * Requires careful hyperparameter tuning (learning rate, tree depth, number of estimators) for optimal performance.

* **Application to AQI**

    * Inputs: Pollutant levels (PM2.5, PM10, CO, NO₂), meteorological data (temperature, humidity, wind).
    * Output: Continuous numeric AQI values or categorical AQI grades.

**Random Forest Regression**

* **Strengths**

    * Combines multiple decision trees to improve prediction stability and reduce variance.
    * Handles non-linear relationships between pollutants and AQI well.
    * Robust against noise and small data variations due to bagging.

* **Weaknesses**

    * Slower for very large datasets compared to boosting methods.
    * Lacks gradient-based optimization, so may not reach the predictive accuracy of XGBoost.
    * Feature importance is less precise for complex datasets.

* **Application to AQI**

    * Inputs and outputs are the same as XGBoost, but predictions may be slightly less accurate for highly non-linear patterns.


**Artificial Neural Networks (ANN)**

* **Strengths**

    * Can learn highly non-linear relationships among multiple pollutants and environmental variables.
    * Flexible architecture allows modeling of complex interactions in AQI data.

* **Weaknesses**

    * Requires larger datasets and more computational resources than tree-based methods.
    * Less interpretable than XGBoost or Random Forest; understanding which pollutant drives the prediction is harder.
    * Sensitive to hyperparameter choices (learning rate, layers, neurons).

* **Application to AQI**

    * Inputs: Same pollutant and meteorological data as above.
    * Output: Continuous numeric AQI values; can also classify AQI in categorical levels.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import xgboost as xgb
import optuna
from sklearn.model_selection import train_test_split, cross_val_score
from sklearn.preprocessing import PolynomialFeatures, RobustScaler, MinMaxScaler
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score
from xgboost import XGBRegressor
from sklearn.ensemble import RandomForestRegressor
```

#### Dataset Loading

```python
#Load Dataset
dataset = pd.read_csv("/city_day.csv")

print("Dataset Information:")
print(dataset.info())
```

```python
# Check the total null values
dataset.isnull().sum()
```

#### Data Preprocessing ####

```python
# Remove rows where AQI is missing (target variable)
dataset = dataset.dropna(subset=['AQI'])

# Train / Validation / Test Split
train_data, test_data = train_test_split(dataset, test_size=0.2, random_state=42)
val_data, test_data = train_test_split(test_data, test_size=0.5, random_state=42)
print("Train shape:", train_data.shape)
print("Validation shape:", val_data.shape)
print("Test shape:", test_data.shape)

# Handle Missing Values
numeric_cols = train_data.select_dtypes(include=['int64', 'float64']).columns.drop('AQI')
train_means = train_data[numeric_cols].mean()
train_data[numeric_cols] = train_data[numeric_cols].fillna(train_means)
val_data[numeric_cols] = val_data[numeric_cols].fillna(train_means)
test_data[numeric_cols] = test_data[numeric_cols].fillna(train_means)


# Drop Non-Numeric / Non-Feature Columns
drop_columns = ['City', 'Date', 'AQI_Bucket']
train_data = train_data.drop(columns=drop_columns)
val_data = val_data.drop(columns=drop_columns)
test_data = test_data.drop(columns=drop_columns)
```

#### Correlation Analysis and Feature Selection

```python
# Correlation Analysis
numeric_train_data = train_data.select_dtypes(include=['int64', 'float64'])
correlation_matrix = numeric_train_data.corr()
correlation_with_aqi = correlation_matrix['AQI'].abs().sort_values(ascending=False)

# Feature Selection
selected_features = correlation_with_aqi[correlation_with_aqi > 0.1].index.drop('AQI').tolist()
print("\nSelected Features based on correlation threshold:")
print(selected_features)


# Correlation Heatmap for Selected Features
filtered_corr_matrix = correlation_matrix.loc[selected_features, selected_features]
plt.figure(figsize=(12, 8))
sns.heatmap(filtered_corr_matrix, annot=True, fmt=".2f", cmap="coolwarm", linewidths=0.5)
plt.title("Correlation Heatmap")
plt.show()
```

#### Correlation Bar Plot

```python
correlation_bar = correlation_with_aqi.drop('AQI')
sorted_correlations = correlation_bar.sort_values(ascending=False)
plt.figure(figsize=(12, 6))
sorted_correlations.plot(kind='bar')
plt.title("Feature Correlations with AQI")
plt.ylabel("Absolute Correlation")
plt.xlabel("Features")
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

#### Input and target features definiton

```python
# Define input features and target variable
X_train = train_data[selected_features].reset_index(drop=True)
X_val = val_data[selected_features].reset_index(drop=True)
X_test = test_data[selected_features].reset_index(drop=True)

y_train = train_data['AQI'].reset_index(drop=True)
y_val = val_data['AQI'].reset_index(drop=True)
y_test = test_data['AQI'].reset_index(drop=True)

print("Train shape:", X_train.shape)
print("Validation shape:", X_val.shape)
print("Test shape:", X_test.shape)
```

#### Log Transformation

```python
#Log transformation for skewed features
skewed_cols = [col for col in selected_features if X_train[col].skew() > 2]  # Dynamic
for dataset in [X_train, X_val, X_test]:
    dataset[skewed_cols] = np.log1p(dataset[skewed_cols])
```

#### Feature Engineering

```python
# Polynomial features
top_corr_features = correlation_with_aqi.drop('AQI').head(6).index.tolist() # numeric features
poly = PolynomialFeatures(degree=2, interaction_only=True, include_bias=False)

X_train_poly = pd.DataFrame(
    poly.fit_transform(X_train[top_corr_features]),
    columns=poly.get_feature_names_out(top_corr_features)
)
X_val_poly = pd.DataFrame(
    poly.transform(X_val[top_corr_features]),
    columns=poly.get_feature_names_out(top_corr_features)
)
X_test_poly = pd.DataFrame(
    poly.transform(X_test[top_corr_features]),
    columns=poly.get_feature_names_out(top_corr_features)
)

for dataset in [X_train_poly, X_val_poly, X_test_poly]:
    dataset.replace([np.inf, -np.inf], np.nan, inplace=True)
    dataset.fillna(0, inplace=True)

X_train_poly_unscaled = X_train_poly.copy()
X_val_poly_unscaled   = X_val_poly.copy()
X_test_poly_unscaled  = X_test_poly.copy()

scaler = RobustScaler()
X_train_poly = scaler.fit_transform(X_train_poly)
X_val_poly = scaler.transform(X_val_poly)
X_test_poly = scaler.transform(X_test_poly)

print("Original feature shape:", X_train.shape)
print("Enhanced feature shape:", X_train_poly.shape)
```

#### Hyperparameter Tuning

```python

# Define objective function for Optuna
def objective(trial):
    # Suggest hyperparameters for XGBoost
    params = {
        "n_estimators": trial.suggest_int("n_estimators", 50, 200),
        "max_depth": trial.suggest_int("max_depth", 3, 6),
        "learning_rate": trial.suggest_float("learning_rate", 0.01, 0.5, log=True),
        "subsample": trial.suggest_float("subsample", 0.6, 0.9),
        "reg_alpha": trial.suggest_float("reg_alpha", 0.01, 10.0, log=True),
        "reg_lambda": trial.suggest_float("reg_lambda", 0.01, 10.0, log=True),
        "objective": "reg:squarederror",
        "random_state": 42,
        "n_jobs": -1
    }

    # Create XGBoost regressor with these params
    model = xgb.XGBRegressor(**params)

    # Evaluate with 5-fold cross-validation
    scores = cross_val_score(
        model,
        X_val_poly,  # training features
        y_val,       # training target
        cv=5,
        scoring="neg_root_mean_squared_error"
    )
    return -np.mean(scores)

study = optuna.create_study(direction="minimize")
study.optimize(objective, n_trials=30)

best_params = study.best_params
best_score = study.best_value
print("Best Parameters:", best_params)
print("Best RMSE:", best_score)

def rf_objective(trial):
    params = {
        "n_estimators": trial.suggest_int("n_estimators", 50, 200),
        "max_depth": trial.suggest_int("max_depth", 3, 6),
        "min_samples_split": trial.suggest_int("min_samples_split", 2, 20),
        "min_samples_leaf": trial.suggest_int("min_samples_leaf", 1, 10),
        "random_state": 42,
        "n_jobs": -1
    }
    model = RandomForestRegressor(**params)
    scores = cross_val_score(model, X_val_poly, y_val, cv=5, scoring="neg_root_mean_squared_error")
    return -np.mean(scores)

rf_study = optuna.create_study(direction="minimize")
rf_study.optimize(rf_objective, n_trials=30)
rf_best_params = rf_study.best_params
print("Best RF Parameters:", rf_best_params)
```

#### Model Training

```python
# =========
#  XGBoost
# =========

xgb_model = XGBRegressor(
    **best_params,                     # Best hyperparameters from Optuna
    objective="reg:squarederror",
    random_state=42,
    n_jobs=-1
)

xgb_model.fit(
    X_train_poly, y_train,
    eval_set=[(X_val_poly, y_val)],  # Use validation for monitoring
    verbose=False
)
y_pred_xgb = xgb_model.predict(X_test_poly)

xgb_mae  = mean_absolute_error(y_test, y_pred_xgb)
xgb_rmse = np.sqrt(mean_squared_error(y_test, y_pred_xgb))
xgb_r2   = r2_score(y_test, y_pred_xgb)

# =========================
#  Random Forest Regressor
# =========================

rf_model = RandomForestRegressor(**rf_best_params)
rf_model.fit(X_train_poly, y_train)
y_pred_rf = rf_model.predict(X_test_poly)

rf_mae  = mean_absolute_error(y_test, y_pred_rf)
rf_rmse = np.sqrt(mean_squared_error(y_test, y_pred_rf))
rf_r2   = r2_score(y_test, y_pred_rf)

# ==========
#  Ensemble
# ==========

y_val_pred_xgb = xgb_model.predict(X_val_poly)
y_val_pred_rf = rf_model.predict(X_val_poly)
val_rmse_xgb = np.sqrt(mean_squared_error(y_val, y_val_pred_xgb))
val_rmse_rf = np.sqrt(mean_squared_error(y_val, y_val_pred_rf))

# Weights: Higher for better model (normalize inverse errors)
weights = [1/val_rmse_xgb, 1/val_rmse_rf]
weights = [w / sum(weights) for w in weights]  # Normalize to sum=1
print(f"Ensemble Weights: XGBoost={weights[0]:.2f}, RF={weights[1]:.2f}")

# VotingRegressor for ensemble (fits if needed, but since trained, use manual weighted pred)
ensemble_pred = (weights[0] * y_pred_xgb + weights[1] * y_pred_rf)
ensemble_mae = mean_absolute_error(y_test, ensemble_pred)
ensemble_rmse = np.sqrt(mean_squared_error(y_test, ensemble_pred))
ensemble_r2 = r2_score(y_test, ensemble_pred)
```

# Testing and Evaluation
---

To verify the correctness and effectiveness of the proposed air quality prediction system, the following testing and evaluation techniques have been used:

+ A hold-out validation strategy was used, splitting the dataset into training (80%), validation (10%), and test (10%) sets. The training set was used to learn model parameters. The validation set was used for hyperparameter tuning and ensemble weighting and the test set was used for final performance evaluation ensuring that the reported results refelct true generalisation/
+ Quantitative metrics: RMSE, MAE, and R² were used to comapre expected and produced results given AQI is a continuous numerical value where both average error and sensitivity to large deviations are important.
+ Visual evaluation: Actual vs. predicted AQI plots and feature importance analysis to confirm logical prediction trends.
+ Cross-validation was also implicitly supported through hyperparameter optimization with Optuna, which uses repeated trials to ensure stable performance.

#### Quantitative Performance Evaluation

```python
results_df = pd.DataFrame({
    "Model": ["XGBoost", "Random Forest", "Ensemble (XGBoost + RF)" ],
    "MAE":  [xgb_mae, rf_mae, ensemble_mae],
    "RMSE": [xgb_rmse, rf_rmse, ensemble_rmse],
    "R²":   [xgb_r2, rf_r2, ensemble_r2]
})

print("Model Performance Comparison:")
display(results_df)
```

As you can see above the ensemble achieved the best overall performance, on the unseen test data, achieving the lowest RMSE value and a highest R² score. This indicates that the model successfully captures complex relationships between multiple pollutant features and air quality levels. The ensemble approach improved robustness by combining predictions from multiple models, reducing individual model bias.

#### Feature Importance Analysis

```python
# Get feature importanc from trained models
importance_xgb = xgb_model.feature_importances_
importance_rf = rf_model.feature_importances_

# Combine (average; align on same features)
features = X_train_poly_unscaled.columns.tolist()
combined_importance = {
    feature: (importance_xgb[i] + importance_rf[i]) / 2
    for i, feature in enumerate(features)
}

feature_importance_df = pd.DataFrame({
    "Feature": list(combined_importance.keys()),
    "Importance": list(combined_importance.values())
}).sort_values(by="Importance", ascending=False).round(4)


plt.figure(figsize=(10, 6))
plt.barh(feature_importance_df.head(10)["Feature"], feature_importance_df.head(10)["Importance"])
plt.gca().invert_yaxis()
plt.title("Top 10 Feature Importances (Ensemble)")
plt.xlabel("Importance Score")
plt.show()
```

The above feature importance analysis revealed that CO / PM2.5, CO / NO and PM2.5 / PM10 ratios had the strongest influence on AQI predictions, while other features contributed less. This aligns with real-world air quality science, where fine particulate matter (PM2.5) and combustion-related gases are known to significantly impact air pollution levels. Furthermore these insightes can be used to prioritise monitoring of PM2.5 and CO levels and improve early warning systems for pollution spikes.

#### Generalization Analaysis

```python
# ---- Training predictions ----
xgb_train_pred = xgb_model.predict(X_train_poly)
rf_train_pred  = rf_model.predict(X_train_poly)

ensemble_train_pred = (
    weights[0] * xgb_train_pred +
    weights[1] * rf_train_pred
)

train_metrics = {
    "MAE": mean_absolute_error(y_train, ensemble_train_pred),
    "RMSE": np.sqrt(mean_squared_error(y_train, ensemble_train_pred))
}

# ---- Testing predictions ----
xgb_test_pred = xgb_model.predict(X_test_poly)
rf_test_pred  = rf_model.predict(X_test_poly)

ensemble_test_pred = (
    weights[0] * xgb_test_pred +
    weights[1] * rf_test_pred
)

test_metrics = {
    "MAE": mean_absolute_error(y_test, ensemble_test_pred),
    "RMSE": np.sqrt(mean_squared_error(y_test, ensemble_test_pred))
}

# ---- Bar plot ----
metrics = ["MAE", "RMSE"]
train_vals = [train_metrics[m] for m in metrics]
test_vals  = [test_metrics[m] for m in metrics]

x = np.arange(len(metrics))
bar_width = 0.35

plt.figure(figsize=(8,5))
plt.bar(x - bar_width/2, train_vals, bar_width, label="Training")
plt.bar(x + bar_width/2, test_vals, bar_width, label="Testing")

plt.xticks(x, metrics)
plt.ylabel("Error")
plt.title("Training vs Testing Performance (Weighted Ensemble Model)")
plt.legend()
plt.grid(axis="y", alpha=0.3)
plt.tight_layout()
plt.show()
```

The above bar plot shows that the training errors is significantly lower than testing errors for both MAE and RMSE. This behaviour is expected in real-world AQI prediction, where unseen environmental conditions introduce additional uncertainty. Importantly, the error gap is not excessive, indicating that the model generalises reasonably well without severe overfitting.
